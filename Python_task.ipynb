### 1. Reverse a String
* **Prompt:** Write a Python program to reverse a given string.

```python
def reverse_string(s):
    return s[::-1]

print(reverse_string("hello"))
```

### 2. Check for Palindrome
* **Prompt:** Write a Python program to check if a string is a palindrome (reads the same backward as forward).

```python
def is_palindrome(s):
    clean_s = "".join(c.lower() for c in s if c.isalnum())
    return clean_s == clean_s[::-1]

print(is_palindrome("A man, a plan, a canal: Panama"))
```

### 3. Find Fibonacci Number (Iterative)
* **Prompt:** Write a Python program to find the $n$-th Fibonacci number efficiently.

```python
def fibonacci(n):
    if n <= 0: return 0
    elif n == 1: return 1
    a, b = 0, 1
    for _ in range(2, n + 1):
        a, b = b, a + b
    return b

print(fibonacci(10))
```

### 5. Find the First Non-Repeating Character
* **Prompt:** Write a Python program to find the first character in a string that does not repeat.

```python
def first_unique_char(s):
    counts = {}
    for c in s:
        counts[c] = counts.get(c, 0) + 1
    for c in s:
        if counts[c] == 1:
            return c
    return None

print(first_unique_char("swiss"))
```

### 6. Merge Two Sorted Lists
* **Prompt:** Write a Python program to merge two pre-sorted lists into a single sorted list.

```python
def merge_lists(l1, l2):
    result = []
    i = j = 0
    while i < len(l1) and j < len(l2):
        if l1[i] < l2[j]:
            result.append(l1[i])
            i += 1
        else:
            result.append(l2[j])
            j += 1
    result.extend(l1[i:])
    result.extend(l2[j:])
    return result

print(merge_lists([1, 3, 5], [2, 4, 6]))
```

### 7. Find Duplicates in a List
* **Prompt:** Write a Python program to find duplicate elements in a list.

```python
def find_duplicates(nums):
    seen = set()
    dups = set()
    for x in nums:
        if x in seen:
            dups.add(x)
        seen.add(x)
    return sorted(list(dups))

print(find_duplicates([1, 2, 3, 2, 4, 5, 1]))
```

### 9. Two Sum Problem
* **Prompt:** Write a Python program to find indices of the two numbers that add up to a target value, given an array of integers.

```python
def two_sum(nums, target):
    mapping = {}
    for i, num in enumerate(nums):
        diff = target - num
        if diff in mapping:
            return [mapping[diff], i]
        mapping[num] = i

print(two_sum([2, 7, 11, 15], 9))
```

### 10. Bubble Sort Implementation
* **Prompt:** Write a Python program to implement the classic bubble sort algorithm.

```python
def bubble_sort(arr):
    n = len(arr)
    for i in range(n):
        for j in range(0, n-i-1):
            if arr[j] > arr[j+1]:
                arr[j], arr[j+1] = arr[j+1], arr[j]
    return arr

print(bubble_sort([64, 34, 25, 12, 22, 11, 90]))
```

### 11. Flatten a Nested List
* **Prompt:** Write a Python program to flatten a deeply nested list of integers into a flat list.

```python
def flatten(nested):
    result = []
    for item in nested:
        if isinstance(item, list):
            result.extend(flatten(item))
        else:
            result.append(item)
    return result

print(flatten([1, [2, [3, 4], 5], 6]))
```

### 12. Check Prime Number
* **Prompt:** Write a Python program to check if a given integer is a prime number.

```python
def is_prime(n):
    if n <= 1: return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0: return False
    return True

print(is_prime(11))
```

### 14. Find the Missing Number in a Sequence
* **Prompt:** Write a Python program to find the missing integer, given a list of $n-1$ unique integers in the range $[1, n]$.

```python
def find_missing(nums, n):
    expected_sum = n * (n + 1) // 2
    actual_sum = sum(nums)
    return expected_sum - actual_sum

print(find_missing([1, 2, 4, 5, 6], 6))
```

### 15. Binary Search
* **Prompt:** Write a Python program to implement a binary search algorithm on a sorted list.

```python
def binary_search(arr, target):
    low, high = 0, len(arr) - 1
    while low <= high:
        mid = (low + high) // 2
        if arr[mid] == target:
            return mid

        elif arr[mid] < target:
            low = mid + 1
        else:
            high = mid - 1
    return -1

print(binary_search([1, 2, 3, 4, 5, 6], 4))
```

### 17. Decorator for Timing Execution
* **Prompt:** Write a Python program to implement a decorator that measures and prints the execution time of a function.

```python
import time

def timer_decorator(func):
    def wrapper(*args, **kwargs):
        start = time.time()
        res = func(*args, **kwargs)
        print(f"{func.__name__} took {time.time() - start:.4f} sec")
        return res
    return wrapper

@timer_decorator
def waste_time():
    time.sleep(0.1)

waste_time()
```

### 18. Generate Subsets (Power Set)
* **Prompt:** Write a Python program to generate all possible subsets (power set) of a given set of numbers.

```python
def power_set(s):
    result = [[]]
    for elem in s:
        result.extend([subset + [elem] for subset in result])
    return result

print(power_set([1, 2]))
```

### 20. Singleton Design Pattern
* **Prompt:** Write a Python program to implement a Singleton pattern using a custom class override.

```python
class Singleton:
    _instance = None
    def __new__(cls):
        if cls._instance is None:
            cls._instance = super(Singleton, cls).__new__(cls)
        return cls._instance

s1 = Singleton()
s2 = Singleton()
print(s1 is s2)
```

### 21. NumPy: Normalize an Array
* **Prompt:** Write a Python program to normalize a 1D NumPy array so that its values range between 0 and 1.

```python
import numpy as np

def normalize_array(arr):
    min_val = np.min(arr)
    max_val = np.max(arr)
    return (arr - min_val) / (max_val - min_val)

input_arr = np.array([10, 20, 30, 40, 50])
print(normalize_array(input_arr))
```

### 25. Pandas: GroupBy and Aggregate
* **Prompt:** Write a Python program to group a DataFrame by a 'Department' column and calculate the average 'Salary' for each department.

```python
import pandas as pd

data = {
    'Department': ['HR', 'IT', 'IT', 'HR', 'Finance'],
    'Salary': [50000, 80000, 95000, 55000, 70000]
}
df = pd.DataFrame(data)

avg_salary = df.groupby('Department')['Salary'].mean().reset_index()
print(avg_salary)
```
